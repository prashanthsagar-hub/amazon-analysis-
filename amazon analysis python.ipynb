```python
df['discounted_price'] = (
    df['discounted_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['discounted_price'] = pd.to_numeric(
    df['discounted_price'], errors='coerce'
)

df['actual_price'] = (
    df['actual_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['actual_price'] = pd.to_numeric(
    df['actual_price'], errors='coerce'
)

df['discount_percentage'] = (
    df['discount_percentage']
    .astype(str)
    .str.replace('%', '', regex=False)
)

df['discount_percentage'] = pd.to_numeric(
    df['discount_percentage'], errors='coerce'
) / 100
```

```python
missing_percentage=df.isnull().sum()/len(df)*100
missing_percentage.plot(kind="bar")
plt.title('Percentage of Missing Values in each Column')
plt.xlabel('Columns')
plt.ylabel("percentage")
plt.show()
```

```python
plt.scatter(df["actual_price"],df["rating"])
plt.title("actual price by rating")
plt.xlabel("actual_price")
plt.ylabel("rating")
plt.show()
```

```python
plt.hist(df["actual_price"])
plt.title("actual price by frequency")
plt.xlabel("actual_price")
plt.ylabel("frequency")
plt.show()
```

```python
a={2:"1-2",3:"2-3",4:"3-4",5:"4-5"}
sizes=list(a.keys())
labels=list(a.values())
plt.pie(sizes,labels=labels,autopct="%1.0f%%")
plt.title("rating by products")
plt.show()
```

```python
a=df.groupby("category")["actual_price"].min().sort_values(ascending=False).head(10)
a.plot(kind="bar")
plt.title("actual price(min) by category")
plt.xlabel("category")
plt.ylabel("actual price")
plt.show()
```

```python
a=df.groupby("category")["actual_price"].max().sort_values(ascending=False).head(10)
a.plot(kind="bar")
plt.title("actual price(max) by category")
plt.xlabel("category")
plt.ylabel("actual price")
plt.show()
```

```python
a=df.groupby("category")["actual_price"].mean().sort_values(ascending=False).head(10)
a.plot(kind="line")
plt.title("actual price(avg) by category")
plt.xlabel("category")
plt.ylabel("actual price")
plt.show()
```

```python
a=df.groupby("category")["discounted_price"].mean().sort_values(ascending=False).head(5)
a.plot(kind="bar")
plt.title("actual price(avg) by category")
plt.xlabel("category")
plt.ylabel("actual price")
plt.show()
```

```python
df["discounted_price"].hist(label="Discounted Price")
df["actual_price"].hist(label="Actual Price")

df["discount_percentage"] = (df["actual_price"] - df["discounted_price"]) / df["actual_price"] * 100
df["discount_percentage"].describe()
df["discount_percentage"].hist(label="Discount Percentage")

plt.title("discounted price by actual price")
plt.xlabel("discounted_price")
plt.ylabel("actual_price")
plt.show()
```
